```python
import pandas as pd
financeData = pd.read_csv('fbc_data_2025.csv')
votingData = pd.read_csv('2024_US_County_Level_Presidential_Results.csv')
```

```python
# Wrangles the financial dataset by taking the 10 rows for each county, averaging their values in each column, and puts the averages into a single row for each county.
# This process also reduces the number of rows for this dataset to 3143
financeDataAveraged = financeData.groupby('county_fips', as_index=False)[['Monthly Housing', 'Monthly Food', 'Monthly Transportation', 'Monthly Healthcare', 'Monthly Other Necessities', 'Monthly Childcare', 'Monthly Taxes', 'Monthly Total',
                                         'Annual Housing', 'Annual Food', 'Annual Transportation', 'Annual Healthcare', 'Annual Other Necessities', 'Annual Childcare', 'Annual Taxes', 'Annual Total',
                                         'median_family_income', 'num_counties_in_st', 'st_cost_rank', 'st_med_aff_rank', 'st_income_rank']].mean()
print(f"Number of rows and columns in financial data: {financeDataAveraged.shape}")
financeDataAveraged.head()
```

```python
print(f"Number of rows and columns in voting data: {votingData.shape}")
votingData.head()
```

```python
# Confirms the use of county_fips as the unit of observation.
print(f"Number of unique counties in financial data: {len(financeDataAveraged['county_fips'].unique())}")
print(f"Number of unique counties in voting data: {len(votingData['county_fips'].unique())}")
```

```python
# Gets the types of variables and non-null counts for the financial data.
financeDataAveraged.info()
```

```python
# Cleans the data by dropping any rows with null values. This is necessary since the info function above showed that some rows have null values.
financeDataAveraged = financeDataAveraged.dropna()
financeDataAveraged.shape
```

```python
# Gets the types of variables and non-null counts for the voting data.
votingData.info()
```

```python
# Gets the summary statistics for the financial data.
financeDataAveraged.describe()
```

```python
# Gets the summary statistics for the voting data.
votingData.describe()
```

```python
# Merges the two datasets together on the county_fips unique column.
combinedData = votingData.merge(financeDataAveraged, how='inner', on='county_fips')
combinedData.head()
```

```python
# Checks that the shape is correct. It should have 31 columns.
# 31 columns comes from 22 columns from the financial data plus 10 columns from the voting data minus 1 common column (county_fips).
# 3104 rows is slightly less than the 3131 and 3160 of the two datasets, but this is intentional as only rows that had their county_fips exist in both datasets were merged.
combinedData.shape
```

```python
# Checks that every county_fips value is unique.
print(f"Number of unique counties in combined data: {len(combinedData['county_fips'].unique())}")
```

```python
# Saving the merged dataset as a CSV file
combinedData.to_csv('merged_dataset.csv', index=False)

# Saving the merged dataset as an Excel file
combinedData.to_excel('merged_dataset.xlsx', index=False)
```
